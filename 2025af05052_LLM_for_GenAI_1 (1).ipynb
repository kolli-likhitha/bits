```python
import os
import glob
from pypdf import PdfReader

# Create output folder
os.makedirs("domain_corpus", exist_ok=True)

# Get all PDFs
pdf_files = glob.glob("datavaol/*.pdf")

# Process PDFs
for pdf_file in pdf_files:

    reader = PdfReader(pdf_file)

    text = ""

    for page in reader.pages:
        page_text = page.extract_text()

        if page_text:
            text += page_text + "\n"

    txt_name = os.path.basename(pdf_file).replace(".pdf", ".txt")

    with open(
        os.path.join("domain_corpus", txt_name),
        "w",
        encoding="utf-8"
    ) as f:
        f.write(text)

    print(f"Saved: {txt_name}")

print("\nDone!")
```

```python
#Length Filtering Statistics
import os
import glob

MIN_LENGTH = 1000

files_before = len(glob.glob("domain_corpus/*.txt"))

removed = 0

for file in glob.glob("domain_corpus/*.txt"):
    with open(file, "r", encoding="utf-8") as f:
        text = f.read()

    if len(text) < MIN_LENGTH:
        os.remove(file)
        removed += 1
        print(f"Removed: {os.path.basename(file)}")

files_after = len(glob.glob("domain_corpus/*.txt"))

print("\nLength Filtering Statistics")
print("-" * 30)
print("Files Before:", files_before)
print("Files Removed:", removed)
print("Files After:", files_after)
```

```python
#Language Filteri*g Statistics
import glob
import os
from langdetect import detect

files_before = len(glob.glob("domain_corpus/*.txt"))

removed = 0

for file in glob.glob("domain_corpus/*.txt"):

    with open(file, "r", encoding="utf-8") as f:
        text = f.read()

    try:
        lang = detect(text)

        if lang != "en":
            os.remove(file)
            removed += 1
            print(f"Removed: {os.path.basename(file)} ({lang})")

    except:
        os.remove(file)
        removed += 1
        print(f"Removed: {os.path.basename(file)} (language detection failed)")

files_after = len(glob.glob("domain_corpus/*.txt"))

print("\nLanguage Filteri*g Statistics")
print("-" * 30)
print("Files Before:", files_before)
print("Files Removed:", removed)
print("Files After :", files_after)
```

```python
#Deduplication Statistics
import glob
import os
import hashlib

files_before = len(glob.glob("domain_corpus/*.txt"))

seen = set()
removed = 0

for file in glob.glob("domain_corpus/*.txt"):

    with open(file, "r", encoding="utf-8") as f:
        text = f.read()

    file_hash = hashlib.md5(
        text.encode("utf-8")
    ).hexdigest()

    if file_hash in seen:
        os.remove(file)
        removed += 1
        print(f"Removed Duplicate: {os.path.basename(file)}")
    else:
        seen.add(file_hash)

files_after = len(glob.glob("domain_corpus/*.txt"))

print("\nDeduplication Statistics")
print("-" * 30)
print("Files Before:", files_before)
print("Duplicates Removed:", removed)
print("Files After:", files_after)
```

The baseline SmolLM2-1.7B-Instruct model generated coherent and grammatically correct responses to HR-related prompts. The model was able to discuss general HR concepts such as leave management, recruitment practices, onboarding procedures, and remote work policies. However, the responses were largely generic and were not grounded in the specific HR policy documents collected in the domain corpus.

For example, when asked about leave policies and remote work guidelines, the model provided broad industry-standard recommendations rather than organization-specific rules and procedures contained in the HR documents. The outputs lacked references to actual policy details, eligibility criteria, approval workflows, and company-specific requirements. Additionally, some responses showed minor repetition and included generic statements that were not directly relevant to the HR corpus.

Overall, the baseline model demonstrated good language generation capabilities but limited domain-specific knowledge of the organization's HR policies. These baseline outputs will serve as a benchmark and will be compared with the QLoRA fine-tuned model in Part B3 to assess improvements in HR policy awareness, factual accuracy, contextual relevance, and domain-specific response quality.

```python
import random
import os

question_templates = [

    "What does this HR policy state?",

    "Summarize this HR policy section.",

    "Explain this employee policy.",

    "What are the key points in this HR guideline?",

    "What should employees know from this section?",

    "Describe the HR process explained here.",

    "What responsibilities are mentioned in this policy?",

    "Explain this workplace guideline.",

    "What requirements are described in this HR policy?",

    "Provide a summary of this section."
]

instruction_dataset = []

for file in os.listdir("domain_corpus"):

    path = os.path.join("domain_corpus", file)

    text = open(
        path,
        encoding="utf-8"
    ).read()

    chunk_size = 500

    for i in range(0, len(text), chunk_size):

        chunk = text[i:i+chunk_size]

        if len(chunk) > 200:

            instruction_dataset.append({

                "instruction":
                random.choice(question_templates),

                "response":
                chunk.strip()

            })

print("Examples:", len(instruction_dataset))
```

The HR policy documents were divided into smaller text chunks of approximately 500 characters each to create instruction-response training examples. For every chunk, an instruction was generated by randomly selecting a question template related to HR policies, guidelines, responsibilities, procedures, and workplace practices. The corresponding text chunk was used as the response.

This approach enabled the automatic creation of a domain-specific instruction dataset from the collected HR documents without manual annotation. The generated dataset contains examples covering various HR topics such as leave policies, recruitment procedures, employee responsibilities, workplace conduct, onboarding processes, and organizational guidelines.

The resulting instruction dataset provides structured input-output pairs suitable for supervised fine-tuning using QLoRA. By training on these HR-specific examples, the model is expected to learn terminology, policy details, and organizational practices present in the domain corpus, thereby improving its ability to answer HR-related queries accurately and contextually.

Dataset Statistics:

Number of source HR documents: 7
Chunk size used: 500 characters
Minimum chunk length retained: 200 characters
Total instruction-response pairs generated: 971

This dataset serves as the foundation for domain adaptation and will be used in subsequent fine-tuning and evaluation stages.

```python
import time

def generate_response(prompt, generation_config):

    formatted_prompt = f"""
Question:
{prompt}

Answer:
"""

    inputs = tokenizer(
        formatted_prompt,
        return_tensors="pt"
    )

    inputs = {
        k: v.to(model.device)
        for k, v in inputs.items()
    }

    start = time.time()

    outputs = model.generate(
        **inputs,
        **generation_config
    )

    end = time.time()

    response = tokenizer.decode(
        outputs[0],
        skip_special_tokens=True
    )

    generation_time = end - start

    return response, generation_time
```

```python
greedy_config = {

    "max_new_tokens": 150,

    "do_sample": False
}
```

```python
for prompt in prompts:

    response, timing = generate_response(
        prompt,
        greedy_config
    )

    print("=" * 80)

    print("PROMPT:")
    print(prompt)

    print("\nTIME (seconds):")
    print(round(timing, 2))

    print("\nRESPONSE:")
    print(response)

    print("\n")
```

```python
# save result
greedy_results = []
for prompt in prompts:
    response, timing = generate_response(
        prompt,
        greedy_config
    )
    greedy_results.append({
        "Prompt": prompt,
        "Strategy": "Greedy",
        "Time": timing,
        "Response": response
    })


# verify result
print(len(greedy_results))

```

```python
# beam search
beam_config = {
    "max_new_tokens": 150,
    "num_beams": 4,
    "do_sample": False
}

# run all 3 prompt
beam_results = []
for prompt in prompts:
    response, timing = generate_response(
        prompt,
        beam_config
    )
    beam_results.append({
        "Prompt": prompt,
        "Strategy": "Beam Search",
        "Time": timing,
        "Response": response
    })
    print("=" * 80)
    print("PROMPT:")
    print(prompt)
    print("\nTIME (seconds):")
    print(round(timing, 2))
    print("\nRESPONSE:")
    print(response)
    print("\n")

# verify results

print("Total Beam Results:", len(beam_results))

# save result
import csv

with open("filename.csv", "w", newline="", encoding="utf-8") as f:
    writer = csv.writer(f)

```

```python
from transformers import BitsAndBytesConfig

bnb_config = BitsAndBytesConfig(
    load_in_4bit=True,
    bnb_4bit_quant_type="nf4",
    bnb_4bit_compute_dtype=torch.bfloat16
)
```

```python
quant_model = AutoModelForCausalLM.from_pretrained(
    MODEL_ID,
    quantization_config=bnb_config,
    device_map="auto"
)
```

Measure VRAM

```python
import torch

torch.cuda.reset_peak_memory_stats()

inputs = tokenizer(
    prompts[0],
    return_tensors="pt"
)

inputs = {
    k: v.to(quant_model.device)
    for k, v in inputs.items()
}

outputs = quant_model.generate(
    **inputs,
    max_new_tokens=150
)

peak_vram = (
    torch.cuda.max_memory_allocated()
    / 1024**3
)

print(
    "Peak VRAM (GB):",
    round(peak_vram, 2)
)
```

Measure Throughput

```python
import time

start = time.time()

outputs = quant_model.generate(
    **inputs,
    max_new_tokens=150
)

end = time.time()

seconds = end - start

throughput = 150 / seconds

print(
    "Generation Time:",
    round(seconds, 2)
)

print(
    "Throughput (tok/s):",
    round(throughput, 2)
)
```

Cost per 1 M Token

```python
hourly_rate = 3.50

cost_per_million = (
    (1000000 / throughput)
    / 3600
    * hourly_rate
)

print(
    "Cost Per 1M Tokens (₹):",
    round(cost_per_million, 2)
)
```
